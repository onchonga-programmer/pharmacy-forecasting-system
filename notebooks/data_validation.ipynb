```python
#import libraries
import sys
import os

# Add project root to Python path
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

import pandas as pd
from sqlalchemy import create_engine
from config.database_config import DATABASE_URL
import matplotlib.pyplot as plt
import seaborn as sns

# Create SQLAlchemy engine 
engine = create_engine(DATABASE_URL)
print('engine created successfully!')
```

```python
# Check what data you have
query = """
SELECT COUNT(*) as total_records FROM raw_data.sales_weekly_raw;
"""
print(pd.read_sql(query, engine))
```

```python
 #Check for missing data
query = """
SELECT 
    year,
    COUNT(*) as weeks_count
FROM raw_data.sales_weekly_raw
GROUP BY year
ORDER BY year;
"""
print(pd.read_sql(query, engine))

# 4. Load data for analysis
query = """
SELECT 
    sw.week_start_date,
    sw.year,
    sw.week_number,
    sw.total_quantity,
    d.drug_name,
    atc.atc_name as category
FROM raw_data.sales_weekly_raw sw
JOIN raw_data.drugs_raw d ON sw.drug_id = d.drug_id
JOIN atc_categories atc ON d.atc_code = atc.atc_code
ORDER BY sw.week_start_date;
"""
df = pd.read_sql(query, engine)

print(f"Total records: {len(df)}")
print(df.head())

# Dispose of the engine (correct method for SQLAlchemy engines)
engine.dispose()
```
